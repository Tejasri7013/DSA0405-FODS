```python
import numpy as np

# Rows: Students; Columns: Math, Science, English, History
student_scores = np.array([
    [85, 90, 78, 88],
    [92, 86, 89, 80],
    [76, 95, 84, 91],
    [88, 89, 92, 85]
])

subjects = ["Math", "Science", "English", "History"]

# Calculate average score for each subject
averages = np.mean(student_scores, axis=0)

# Display average scores
for subject, avg in zip(subjects, averages):
    print(f"Average score in {subject}: {avg:.2f}")

# Find the subject with the highest average
highest_index = np.argmax(averages)

print("\nSubject with highest average:", subjects[highest_index])
print("Highest average score:", averages[highest_index])
```

```python
import numpy as np

# Each row represents a product's sales prices
sales_data = np.array([
    [100, 150, 200],
    [250, 300, 350],
    [120, 180, 240]
])

# Calculate average price of all products sold
average_price = np.mean(sales_data)

print("Average price of all products sold:", average_price)
```

```python
import numpy as np

# Columns: Bedrooms, Square Footage, Sale Price
house_data = np.array([
    [3, 1500, 200000],
    [5, 2500, 350000],
    [4, 2000, 280000],
    [6, 3000, 450000],
    [5, 2200, 400000]
])

# Select houses with more than 4 bedrooms
selected_houses = house_data[house_data[:, 0] > 4]

# Calculate average sale price (third column)
if len(selected_houses) > 0:
    average_price = np.mean(selected_houses[:, 2])
    print("Average sale price of houses with more than 4 bedrooms:",
          average_price)
else:
    print("No houses found with more than 4 bedrooms.")
```

```python
import numpy as np

# Sales for Q1, Q2, Q3, and Q4
sales_data = np.array([10000, 12000, 15000, 18000])

# Calculate total annual sales
total_sales = np.sum(sales_data)

# Calculate percentage increase from Q1 to Q4
if sales_data[0] != 0:
    percentage_increase = (
        (sales_data[3] - sales_data[0]) / sales_data[0]
    ) * 100
else:
    percentage_increase = None

print("Total annual sales:", total_sales)

if percentage_increase is not None:
    print(f"Percentage increase from Q1 to Q4: {percentage_increase:.2f}%")
else:
    print("Percentage increase cannot be calculated because Q1 sales are zero.")
```

```python
import numpy as np

# Fuel efficiency of different car models in miles per gallon (MPG)
fuel_efficiency = np.array([25, 30, 35, 40, 45])

# Calculate average fuel efficiency
average_efficiency = np.mean(fuel_efficiency)

# Compare two car models
model1 = fuel_efficiency[0]  # First car: 25 MPG
model2 = fuel_efficiency[3]  # Fourth car: 40 MPG

# Calculate percentage improvement
if model1 != 0:
    improvement = ((model2 - model1) / model1) * 100
else:
    improvement = None

print(f"Average fuel efficiency: {average_efficiency:.2f} MPG")

if improvement is not None:
    print(f"Percentage improvement: {improvement:.2f}%")
else:
    print("Percentage improvement cannot be calculated because the first model has zero MPG.")
```

```python
# Item prices and quantities
prices = [100, 200, 150, 50]
quantities = [2, 1, 3, 4]

# Discount and tax rates
discount_rate = 10  # 10%
tax_rate = 5        # 5%

# Calculate subtotal
subtotal = sum(
    price * quantity
    for price, quantity in zip(prices, quantities)
)

# Calculate discount
discount = subtotal * (discount_rate / 100)

# Subtotal after discount
discounted_total = subtotal - discount

# Calculate tax on discounted total
tax = discounted_total * (tax_rate / 100)

# Final cost
total_cost = discounted_total + tax

print(f"Subtotal: Rs. {subtotal:.2f}")
print(f"Discount: Rs. {discount:.2f}")
print(f"Amount after discount: Rs. {discounted_total:.2f}")
print(f"Tax: Rs. {tax:.2f}")
print(f"Final total cost: Rs. {total_cost:.2f}")
```
